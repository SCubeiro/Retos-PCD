````python
# PARTE 1: Carga y Exploración de Datos

def cargar_datos() -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """
    Carga los datos de estudiantes, calificaciones y materias.
    """
    # Datos de estudiantes
    estudiantes = pd.DataFrame({
        'boleta': ['2021630001', '2021630002', '2021630003', '2021630004', '2021630005',
                   '2022630001', '2022630002', '2022630003', '2022630004', '2022630005',
                   '2023630001', '2023630002', '2023630003', '2023630004', '2023630005'],
        'nombre': ['Juan Pérez García', 'María López Ruiz', 'Pedro Sánchez Torres',
                   'Ana Martínez Díaz', 'Luis Rodríguez Vega', 'Carmen Flores Luna',
                   'Roberto Díaz Mora', 'Laura Torres Silva', 'Diego Ramírez Cruz',
                   'Sofía Vargas Romo', 'Carlos Mendoza Ríos', 'Patricia Ortiz León',
                   'Miguel Ángel Castro', 'Fernanda Reyes Paz', 'Andrés Guzmán Villa'],
        'semestre': [4, 4, 4, 4, 4, 3, 3, 3, 3, 3, 2, 2, 2, 2, 2],
        'carrera': ['CD'] * 15,
        'email': ['juan.perez@ipn.mx', 'maria.lopez@ipn.mx', 'pedro.sanchez@ipn.mx',
                  'ana.martinez@ipn.mx', 'luis.rodriguez@ipn.mx', 'carmen.flores@ipn.mx',
                  'roberto.diaz@ipn.mx', 'laura.torres@ipn.mx', 'diego.ramirez@ipn.mx',
                  'sofia.vargas@ipn.mx', 'carlos.mendoza@ipn.mx', 'patricia.ortiz@ipn.mx',
                  'miguel.castro@ipn.mx', 'fernanda.reyes@ipn.mx', 'andres.guzman@ipn.mx']
    })
    
    # Datos de materias
    materias = pd.DataFrame({
        'materia_id': ['MAT101', 'MAT102', 'PROG101', 'PROG102', 'EST101', 'EST102', 'BD101'],
        'nombre': ['Cálculo Diferencial', 'Cálculo Integral', 'Programación I',
                   'Programación II', 'Probabilidad', 'Estadística Inferencial',
                   'Bases de Datos'],
        'creditos': [8, 8, 6, 6, 6, 6, 6],
        'semestre_materia': [1, 2, 1, 2, 2, 3, 3]
    })
    
    # Generar calificaciones (simuladas)
    np.random.seed(42)
    calificaciones_data = []
    
    for boleta in estudiantes['boleta']:
        semestre = estudiantes[estudiantes['boleta'] == boleta]['semestre'].values[0]
        materias_cursadas = materias[materias['semestre_materia'] <= semestre]['materia_id'].tolist()
        
        for materia in materias_cursadas:
            # Generar calificaciones aleatorias (con algunos casos especiales)
            base = np.random.uniform(5, 10)
            p1 = round(min(10, max(0, base + np.random.normal(0, 1))), 1)
            p2 = round(min(10, max(0, base + np.random.normal(0, 1))), 1)
            final = round(min(10, max(0, base + np.random.normal(0, 0.5))), 1)
            
            # Algunos valores nulos aleatorios
            if np.random.random() < 0.05:
                p2 = np.nan
            
            calificaciones_data.append({
                'boleta': boleta,
                'materia_id': materia,
                'parcial_1': p1,
                'parcial_2': p2,
                'final': final
            })
    
    calificaciones = pd.DataFrame(calificaciones_data)
    
    return estudiantes, calificaciones, materias
````

````python
# Funciones auxiliares de uso interno para los cálculos de promedios.

# Columnas de calificación que componen el promedio de cada materia.
COLS_CALIF = ["parcial_1", "parcial_2", "final"]


def _promedio_por_materia(df_calif: pd.DataFrame) -> pd.Series:
    """
    Promedio de cada registro (materia de un estudiante) como la media de
    parcial_1, parcial_2 y final IGNORANDO NaN (mean(axis=1) hace skipna).
    Así una materia con un parcial faltante promedia los parciales presentes.
    """
    return df_calif[COLS_CALIF].mean(axis=1)


def _promedio_por_estudiante(df_calif: pd.DataFrame) -> pd.Series:
    """
    Promedio general por estudiante: media de los promedios de sus materias.
    Devuelve una Series indexada por boleta.
    """
    tmp = df_calif.copy()
    tmp["promedio"] = _promedio_por_materia(tmp)
    return tmp.groupby("boleta")["promedio"].mean()
````

````python
def info_general(df_estudiantes: pd.DataFrame, df_calificaciones: pd.DataFrame) -> Dict:
    """
    Genera información general del sistema.
    """
    return {
        "total_estudiantes": int(len(df_estudiantes)),
        "total_registros_calif": int(len(df_calificaciones)),
        # Lista de semestres distintos presentes, ordenada.
        "semestres": sorted(df_estudiantes["semestre"].dropna().unique().tolist()),
        "materias_con_registros": int(df_calificaciones["materia_id"].nunique()),
    }
````

````python
def validar_datos(df_calificaciones: pd.DataFrame) -> Dict:
    """
    Valida la integridad de las calificaciones.

    - registros_con_nulos: filas con al menos un valor faltante.
    - calificaciones_fuera_rango: cuántas notas individuales (en parcial_1,
      parcial_2 y final) están fuera de [0, 10]; los NaN no se cuentan.
    - datos_validos: True solo si no hay nulos NI valores fuera de rango.
    """
    cols = [c for c in COLS_CALIF if c in df_calificaciones.columns]

    registros_con_nulos = int(df_calificaciones.isna().any(axis=1).sum())

    notas = df_calificaciones[cols]
    # Las comparaciones con NaN dan False, así que los nulos no se cuentan aquí.
    fuera_rango = int(((notas < 0) | (notas > 10)).sum().sum())

    datos_validos = bool(registros_con_nulos == 0 and fuera_rango == 0)

    return {
        "registros_con_nulos": registros_con_nulos,
        "calificaciones_fuera_rango": fuera_rango,
        "datos_validos": datos_validos,
    }
````

````python
# PARTE 2: Consultas y Filtros

def buscar_estudiante(df_estudiantes: pd.DataFrame, criterio: str, valor: str) -> pd.DataFrame:
    """
    Busca estudiantes por 'boleta' (coincidencia exacta), 'nombre' (parcial, sin
    distinguir mayúsculas) o 'semestre' (igualdad numérica).

    Un criterio no reconocido o un semestre no numérico lanzan ValueError con un
    mensaje claro (no se falla en silencio).
    """
    if criterio == "boleta":
        return df_estudiantes[df_estudiantes["boleta"] == str(valor)]

    if criterio == "nombre":
        return df_estudiantes[
            df_estudiantes["nombre"].str.contains(str(valor), case=False, na=False)
        ]

    if criterio == "semestre":
        try:
            sem = int(valor)
        except (ValueError, TypeError):
            raise ValueError(
                f"El semestre debe ser un número entero; se recibió: {valor!r}"
            )
        return df_estudiantes[df_estudiantes["semestre"] == sem]

    raise ValueError(
        f"Criterio no válido: {criterio!r}. Use 'boleta', 'nombre' o 'semestre'."
    )
````

````python
def obtener_kardex(boleta: str, df_estudiantes: pd.DataFrame,
                   df_calificaciones: pd.DataFrame, df_materias: pd.DataFrame) -> Dict:
    """
    Obtiene el kardex completo de un estudiante.

    Si la boleta no existe, 'estudiante' queda en None (la visualización ya
    contempla ese caso). El promedio de cada materia ignora parciales faltantes.
    """
    resultado = {
        "estudiante": None,
        "materias": None,
        "promedio_general": 0.0,
        "creditos_cursados": 0,
        "materias_aprobadas": 0,
        "materias_reprobadas": 0,
    }

    fila = df_estudiantes[df_estudiantes["boleta"] == str(boleta)]
    if fila.empty:
        return resultado  # estudiante no encontrado
    resultado["estudiante"] = fila.iloc[0].to_dict()

    calif = df_calificaciones[df_calificaciones["boleta"] == str(boleta)].copy()
    if calif.empty:
        resultado["materias"] = calif  # DataFrame vacío, sin calificaciones
        return resultado

    # Unir con el nombre y los créditos de cada materia.
    calif = calif.merge(
        df_materias[["materia_id", "nombre", "creditos"]],
        on="materia_id", how="left",
    )
    calif["promedio"] = _promedio_por_materia(calif).round(2)

    # Tabla legible de materias para el kardex.
    resultado["materias"] = calif[
        ["materia_id", "nombre", "parcial_1", "parcial_2", "final", "promedio"]
    ].rename(columns={"nombre": "materia"})

    promedios = calif["promedio"]
    resultado["promedio_general"] = (
        float(promedios.mean()) if promedios.notna().any() else 0.0
    )
    resultado["creditos_cursados"] = int(calif["creditos"].fillna(0).sum())
    # Las materias con promedio NaN no cuentan ni como aprobadas ni reprobadas.
    resultado["materias_aprobadas"] = int((promedios >= 6).sum())
    resultado["materias_reprobadas"] = int((promedios < 6).sum())

    return resultado
````

````python
def filtrar_por_rendimiento(df_calificaciones: pd.DataFrame,
                            df_estudiantes: pd.DataFrame,
                            min_promedio: float = None,
                            max_promedio: float = None) -> pd.DataFrame:
    """
    Filtra estudiantes cuyo promedio general cae en el rango [min, max].
    Cualquier límite en None se ignora.
    """
    promedios = _promedio_por_estudiante(df_calificaciones).reset_index()
    promedios.columns = ["boleta", "promedio"]
    promedios["promedio"] = promedios["promedio"].round(2)

    if min_promedio is not None:
        promedios = promedios[promedios["promedio"] >= min_promedio]
    if max_promedio is not None:
        promedios = promedios[promedios["promedio"] <= max_promedio]

    resultado = promedios.merge(
        df_estudiantes[["boleta", "nombre", "semestre"]], on="boleta", how="left"
    )
    return (
        resultado[["boleta", "nombre", "semestre", "promedio"]]
        .sort_values("promedio", ascending=False)
        .reset_index(drop=True)
    )
````

````python
# PARTE 3: Cálculos y Estadísticas

def calcular_promedio_materia(df_calificaciones: pd.DataFrame, materia_id: str) -> Dict:
    """
    Calcula estadísticas de una materia.

    'calificacion_maxima'/'calificacion_minima' se calculan sobre el promedio por
    alumno de esa materia. Aprobado = promedio de la materia >= 6.
    """
    resultado = {
        "materia": materia_id,
        "inscritos": 0,
        "promedio_parcial1": None,
        "promedio_parcial2": None,
        "promedio_final": None,
        "promedio_general": None,
        "tasa_aprobacion": None,
        "calificacion_maxima": None,
        "calificacion_minima": None,
    }

    datos = df_calificaciones[df_calificaciones["materia_id"] == materia_id]
    if datos.empty:
        return resultado  # materia sin registros

    promedios = _promedio_por_materia(datos)

    resultado["inscritos"] = int(len(datos))
    resultado["promedio_parcial1"] = float(datos["parcial_1"].mean())
    resultado["promedio_parcial2"] = float(datos["parcial_2"].mean())
    resultado["promedio_final"] = float(datos["final"].mean())
    resultado["promedio_general"] = float(promedios.mean())
    aprobados = int((promedios >= 6).sum())
    # len(datos) > 0 aquí, así que no hay división por cero.
    resultado["tasa_aprobacion"] = float(aprobados / len(datos) * 100)
    resultado["calificacion_maxima"] = float(promedios.max())
    resultado["calificacion_minima"] = float(promedios.min())

    return resultado
````

````python
def ranking_estudiantes(df_calificaciones: pd.DataFrame,
                        df_estudiantes: pd.DataFrame,
                        top_n: int = 10) -> pd.DataFrame:
    """
    Genera el ranking de mejores estudiantes por promedio general (descendente).
    Columnas: Posición, Nombre, Semestre, Promedio.
    """
    promedios = _promedio_por_estudiante(df_calificaciones).reset_index()
    promedios.columns = ["boleta", "promedio"]

    tabla = (
        promedios.merge(
            df_estudiantes[["boleta", "nombre", "semestre"]], on="boleta", how="left"
        )
        .sort_values("promedio", ascending=False)
        .reset_index(drop=True)
    )

    # top_n <= 0 -> ranking vacío; en otro caso se toman los primeros top_n.
    if top_n is not None:
        tabla = tabla.head(top_n) if top_n > 0 else tabla.iloc[0:0]

    # Se usan .values para alinear por posición (1..n) y no por índice.
    return pd.DataFrame({
        "Posición": range(1, len(tabla) + 1),
        "Nombre": tabla["nombre"].values,
        "Semestre": tabla["semestre"].values,
        "Promedio": tabla["promedio"].round(2).values,
    })
````

````python
def estadisticas_por_semestre(df_estudiantes: pd.DataFrame,
                              df_calificaciones: pd.DataFrame) -> pd.DataFrame:
    """
    Estadísticas agrupadas por semestre del estudiante.

    Columnas: Estudiantes, Promedio (media de los promedios por alumno),
    Tasa Aprob. (% de registros materia-alumno con promedio >= 6),
    Mejor y Peor (mejor/peor promedio de alumno del semestre).
    """
    # Promedio de cada registro materia-alumno, con el semestre del estudiante.
    calif = df_calificaciones.copy()
    calif["promedio"] = _promedio_por_materia(calif)
    calif = calif.merge(df_estudiantes[["boleta", "semestre"]], on="boleta", how="left")

    # Promedio general por alumno, con su semestre.
    prom_alumno = _promedio_por_estudiante(df_calificaciones).reset_index()
    prom_alumno.columns = ["boleta", "promedio_alumno"]
    prom_alumno = prom_alumno.merge(
        df_estudiantes[["boleta", "semestre"]], on="boleta", how="left"
    )

    filas = []
    for semestre, grupo in prom_alumno.groupby("semestre"):
        registros = calif[calif["semestre"] == semestre]["promedio"].dropna()
        tasa = (registros >= 6).mean() * 100 if len(registros) else 0.0
        filas.append({
            "Semestre": semestre,
            "Estudiantes": int(len(grupo)),
            "Promedio": round(float(grupo["promedio_alumno"].mean()), 2),
            "Tasa Aprob.": round(float(tasa), 1),
            "Mejor": round(float(grupo["promedio_alumno"].max()), 2),
            "Peor": round(float(grupo["promedio_alumno"].min()), 2),
        })

    resultado = pd.DataFrame(filas)
    if not resultado.empty:
        resultado = resultado.set_index("Semestre").sort_index()
    return resultado
````

````python
# PARTE 4: Identificación de Riesgo y Reportes

def identificar_estudiantes_riesgo(df_calificaciones: pd.DataFrame,
                                   df_estudiantes: pd.DataFrame,
                                   umbral_promedio: float = 7.0,
                                   max_reprobadas: int = 2) -> pd.DataFrame:
    """
    Identifica estudiantes en riesgo académico.

    En riesgo = promedio general < umbral_promedio O más de max_reprobadas
    materias reprobadas (promedio de materia < 6). La columna 'Motivo' indica la
    causa: 'Bajo promedio', 'Mat. reprob.' o 'Ambos'.
    """
    calif = df_calificaciones.copy()
    calif["promedio"] = _promedio_por_materia(calif)

    # Promedio general y número de materias reprobadas por estudiante.
    prom = calif.groupby("boleta")["promedio"].mean()
    reprob = (
        calif.assign(reprobada=calif["promedio"] < 6)
        .groupby("boleta")["reprobada"].sum().astype(int)
    )
    resumen = pd.DataFrame({"promedio": prom, "reprobadas": reprob}).reset_index()

    bajo = resumen["promedio"] < umbral_promedio
    muchas = resumen["reprobadas"] > max_reprobadas
    resumen = resumen[bajo | muchas].copy()

    columnas = ["Boleta", "Nombre", "Promedio", "Reprobadas", "Motivo"]
    if resumen.empty:
        return pd.DataFrame(columns=columnas)  # nadie en riesgo

    def _motivo(row):
        b = row["promedio"] < umbral_promedio
        m = row["reprobadas"] > max_reprobadas
        if b and m:
            return "Ambos"
        return "Bajo promedio" if b else "Mat. reprob."

    resumen["Motivo"] = resumen.apply(_motivo, axis=1)
    resumen = resumen.merge(df_estudiantes[["boleta", "nombre"]], on="boleta", how="left")

    return pd.DataFrame({
        "Boleta": resumen["boleta"].values,
        "Nombre": resumen["nombre"].values,
        "Promedio": resumen["promedio"].round(2).values,
        "Reprobadas": resumen["reprobadas"].astype(int).values,
        "Motivo": resumen["Motivo"].values,
    }).sort_values("Promedio").reset_index(drop=True)
````

````python
def generar_reporte_academico(df_estudiantes: pd.DataFrame,
                              df_calificaciones: pd.DataFrame,
                              df_materias: pd.DataFrame) -> Dict:
    """
    Genera el reporte académico completo integrando las funciones anteriores.
    """
    reporte = {
        "resumen_general": {},
        "por_semestre": None,
        "por_materia": None,
        "mejores_estudiantes": None,
        "estudiantes_riesgo": None,
        "fecha_generacion": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }

    calif = df_calificaciones.copy()
    calif["promedio"] = _promedio_por_materia(calif)

    prom_alumno = _promedio_por_estudiante(df_calificaciones)
    promedio_global = float(prom_alumno.mean()) if len(prom_alumno) else 0.0
    # Tasa global: % de registros materia-alumno aprobados (promedio >= 6).
    tasa_global = float((calif["promedio"] >= 6).mean() * 100) if len(calif) else 0.0

    reporte["resumen_general"] = {
        "total_estudiantes": int(len(df_estudiantes)),
        "promedio_global": round(promedio_global, 2),
        "tasa_aprobacion": round(tasa_global, 1),
    }
    reporte["por_semestre"] = estadisticas_por_semestre(df_estudiantes, df_calificaciones)

    # Una fila por materia con sus estadísticas.
    filas_materia = [calcular_promedio_materia(df_calificaciones, m)
                     for m in df_materias["materia_id"]]
    reporte["por_materia"] = pd.DataFrame(filas_materia)

    reporte["mejores_estudiantes"] = ranking_estudiantes(df_calificaciones, df_estudiantes, top_n=10)
    reporte["estudiantes_riesgo"] = identificar_estudiantes_riesgo(df_calificaciones, df_estudiantes)

    return reporte
````

````python
def exportar_kardex(boleta: str, kardex: Dict, formato: str = 'csv') -> str:
    """
    Exporta el kardex a un archivo CSV o JSON y devuelve el nombre del archivo.

    Lanza ValueError si el estudiante no existe en el kardex o si el formato no es
    'csv' ni 'json'.
    """
    import json

    if kardex.get("estudiante") is None:
        raise ValueError("No se puede exportar: el estudiante no existe en el kardex.")

    formato = formato.lower()
    fecha = datetime.now().strftime("%Y%m%d")
    nombre_archivo = f"kardex_{boleta}_{fecha}.{formato}"
    materias = kardex.get("materias")

    if formato == "csv":
        tabla = materias if (materias is not None and len(materias) > 0) else pd.DataFrame()
        tabla.to_csv(nombre_archivo, index=False)
    elif formato == "json":
        contenido = {
            "estudiante": kardex.get("estudiante"),
            "promedio_general": kardex.get("promedio_general"),
            "creditos_cursados": kardex.get("creditos_cursados"),
            "materias_aprobadas": kardex.get("materias_aprobadas"),
            "materias_reprobadas": kardex.get("materias_reprobadas"),
            "materias": materias.to_dict(orient="records") if materias is not None else [],
        }
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            json.dump(contenido, f, ensure_ascii=False, indent=2, default=str)
    else:
        raise ValueError(f"Formato no válido: {formato!r}. Use 'csv' o 'json'.")

    return nombre_archivo
````

### 1. Predictor de riesgo

````python
def predecir_riesgo_proximo_semestre(df_calificaciones: pd.DataFrame,
                                     df_estudiantes: pd.DataFrame) -> pd.DataFrame:
    """
    Predice posible riesgo futuro por tendencia a la baja: estudiantes cuya
    calificación final es menor que su parcial_1 (la nota empeora hacia el final)
    en más de la mitad de sus materias.
    """
    calif = df_calificaciones.copy()
    calif["tendencia_baja"] = calif["final"] < calif["parcial_1"]

    agg = calif.groupby("boleta").agg(
        materias=("materia_id", "count"),
        materias_a_la_baja=("tendencia_baja", "sum"),
    ).reset_index()
    agg["proporcion_baja"] = (agg["materias_a_la_baja"] / agg["materias"]).round(2)

    en_riesgo = agg[agg["proporcion_baja"] > 0.5].merge(
        df_estudiantes[["boleta", "nombre", "semestre"]], on="boleta", how="left"
    )
    return (
        en_riesgo[["boleta", "nombre", "semestre", "materias",
                   "materias_a_la_baja", "proporcion_baja"]]
        .sort_values("proporcion_baja", ascending=False)
        .reset_index(drop=True)
    )
````

### 2. Comparador de rendimiento

````python
def comparar_estudiantes(boleta1: str, boleta2: str,
                         df_calificaciones: pd.DataFrame,
                         df_estudiantes: pd.DataFrame,
                         df_materias: pd.DataFrame) -> Dict:
    """
    Compara el rendimiento de dos estudiantes (promedio, créditos y materias
    aprobadas/reprobadas) e indica quién tiene mejor promedio.
    """
    k1 = obtener_kardex(boleta1, df_estudiantes, df_calificaciones, df_materias)
    k2 = obtener_kardex(boleta2, df_estudiantes, df_calificaciones, df_materias)

    def _resumen(k):
        est = k["estudiante"] or {}
        return {
            "nombre": est.get("nombre"),
            "promedio_general": k["promedio_general"],
            "creditos_cursados": k["creditos_cursados"],
            "materias_aprobadas": k["materias_aprobadas"],
            "materias_reprobadas": k["materias_reprobadas"],
        }

    r1, r2 = _resumen(k1), _resumen(k2)
    if r1["promedio_general"] == r2["promedio_general"]:
        mejor = "empate"
    else:
        mejor = boleta1 if r1["promedio_general"] > r2["promedio_general"] else boleta2

    return {boleta1: r1, boleta2: r2, "mejor_promedio": mejor}
````
