## Especificaciones de Formatos

### 1. Código de Producto
```
Formato: ABC-1234-MX
         ───  ────  ──
          │    │    └── País: 2 letras mayúsculas
          │    └─────── Número: exactamente 4 dígitos
          └──────────── Categoría: exactamente 3 letras mayúsculas

Ejemplos válidos:   TEC-0001-MX, ALI-9999-US, ROB-1234-CA
Ejemplos inválidos: tec-0001-MX, TEC-001-MX, TEC-12345-MX, TEC-0001-M
```

### 2. Código de Envío
```
Formato: ENV-YYYY-MM-DD-NNNNNN
         ───  ────  ──  ──  ──────
          │    │    │   │    └── Número secuencial: 6 dígitos
          │    │    │   └─────── Día: 01-31
          │    │    └──────────── Mes: 01-12
          │    └───────────────── Año: 4 dígitos (2020-2030)
          └────────────────────── Prefijo: siempre "ENV"

Ejemplos válidos:   ENV-2024-03-15-001234, ENV-2025-12-01-999999
Ejemplos inválidos: ENV-2019-03-15-001234, ENV-2024-13-15-001234, ENV-2024-03-32-001234
```

### 3. Código de Empleado
```
Formato: EMP-XXX-NNNN
         ───  ───  ────
          │    │    └── Número: 4 dígitos (no puede empezar con 0)
          │    └─────── Departamento: 3 letras mayúsculas
          └──────────── Prefijo: siempre "EMP"

Departamentos válidos: VEN (Ventas), ADM (Administración), TEC (Tecnología),
                       LOG (Logística), RHH (Recursos Humanos)

Ejemplos válidos:   EMP-VEN-1234, EMP-TEC-9999, EMP-ADM-1000
Ejemplos inválidos: EMP-VEN-0123, EMP-XXX-1234, EMP-VEN-123
```

### 4. Código de Factura
```
Formato: FAC-S-NNNNNN
         ───  ─  ──────
          │   │    └── Número: 6 dígitos
          │   └─────── Serie: A, B, C, D o E
          └──────────── Prefijo: siempre "FAC"

Ejemplos válidos:   FAC-A-123456, FAC-E-000001
Ejemplos inválidos: FAC-F-123456, FAC-A-12345, FAC-a-123456
```

## Código Base

Usa este código como punto de partida:

````python
import re
from typing import Dict, List

# Departamentos válidos para empleados
DEPARTAMENTOS_VALIDOS = ['VEN', 'ADM', 'TEC', 'LOG', 'RHH']

# Series válidas para facturas
SERIES_VALIDAS = ['A', 'B', 'C', 'D', 'E']
````

````python
# PARTE 1: Funciones de validación individual

def validar_producto(codigo: str) -> Dict:
    """
    Valida código de producto.
    Formato: ABC-1234-MX (3 letras mayus - 4 dígitos - 2 letras país mayus)
    """
    resultado = {
        "valido": False,
        "categoria": None,
        "numero": None,
        "pais": None
    }

    # Patron: 3 letras mayus, 4 digitos, 2 letras mayus, separados por guiones.
    # Anclas ^ y $ para que NO acepte texto extra antes o despues.
    patron = r'^([A-Z]{3})-(\d{4})-([A-Z]{2})$'
    match = re.match(patron, codigo)

    if match:
        resultado["valido"] = True
        resultado["categoria"] = match.group(1)
        resultado["numero"] = match.group(2)
        resultado["pais"] = match.group(3)

    return resultado
````

````python
def validar_envio(codigo: str) -> Dict:
    """
    Valida código de envío.
    Formato: ENV-YYYY-MM-DD-NNNNNN
    - Año: 2020-2030
    - Mes: 01-12
    - Día: 01-31
    """
    resultado = {
        "valido": False,
        "fecha": None,
        "secuencial": None
    }

    # El regex valida la forma general; los rangos de año/mes/dia se
    # checan despues con int() para no inflar el patron.
    patron = r'^ENV-(\d{4})-(\d{2})-(\d{2})-(\d{6})$'
    match = re.match(patron, codigo)

    if not match:
        return resultado

    anio = int(match.group(1))
    mes = int(match.group(2))
    dia = int(match.group(3))
    secuencial = match.group(4)

    if not (2020 <= anio <= 2030):
        return resultado
    if not (1 <= mes <= 12):
        return resultado
    if not (1 <= dia <= 31):
        return resultado

    resultado["valido"] = True
    resultado["fecha"] = f"{anio:04d}-{mes:02d}-{dia:02d}"
    resultado["secuencial"] = secuencial

    return resultado
````

````python
def validar_empleado(codigo: str) -> Dict:
    """
    Valida código de empleado.
    Formato: EMP-XXX-NNNN
    - Departamento: debe estar en DEPARTAMENTOS_VALIDOS
    - Número: 4 dígitos, NO empieza con 0
    """
    resultado = {
        "valido": False,
        "departamento": None,
        "numero": None
    }

    # [1-9]\d{3} fuerza que el primer digito sea 1-9 (no empieza con 0).
    patron = r'^EMP-([A-Z]{3})-([1-9]\d{3})$'
    match = re.match(patron, codigo)

    if not match:
        return resultado

    departamento = match.group(1)
    numero = match.group(2)

    # El regex acepta cualquier 3 letras, asi que validamos contra la lista.
    if departamento not in DEPARTAMENTOS_VALIDOS:
        return resultado

    resultado["valido"] = True
    resultado["departamento"] = departamento
    resultado["numero"] = numero

    return resultado
````

````python
def validar_factura(codigo: str) -> Dict:
    """
    Valida código de factura.
    Formato: FAC-S-NNNNNN
    - Serie: A, B, C, D o E
    - Número: 6 dígitos
    """
    resultado = {
        "valido": False,
        "serie": None,
        "numero": None
    }

    # [A-E] limita la serie directamente desde el patron.
    patron = r'^FAC-([A-E])-(\d{6})$'
    match = re.match(patron, codigo)

    if match:
        resultado["valido"] = True
        resultado["serie"] = match.group(1)
        resultado["numero"] = match.group(2)

    return resultado
````

````python
# PARTE 2: Validador universal

def validar_codigo(codigo: str) -> Dict:
    """
    Detecta el tipo de código por su prefijo y delega al validador
    correspondiente.
    """
    resultado = {
        "codigo": codigo,
        "tipo": "desconocido",
        "valido": False,
        "detalles": {}
    }

    # ENV, EMP y FAC se distinguen por prefijo. Los productos no tienen
    # prefijo fijo, asi que se reconocen por su forma general (3 bloques
    # letras-digitos-letras), aunque despues validar_producto rechace los
    # que no cumplan el patron exacto.
    if codigo.startswith("ENV-"):
        detalles = validar_envio(codigo)
        resultado["tipo"] = "envio"
        resultado["valido"] = detalles["valido"]
        if detalles["valido"]:
            resultado["detalles"] = {
                "fecha": detalles["fecha"],
                "secuencial": detalles["secuencial"]
            }
    elif codigo.startswith("EMP-"):
        detalles = validar_empleado(codigo)
        resultado["tipo"] = "empleado"
        resultado["valido"] = detalles["valido"]
        if detalles["valido"]:
            resultado["detalles"] = {
                "departamento": detalles["departamento"],
                "numero": detalles["numero"]
            }
    elif codigo.startswith("FAC-"):
        detalles = validar_factura(codigo)
        resultado["tipo"] = "factura"
        resultado["valido"] = detalles["valido"]
        if detalles["valido"]:
            resultado["detalles"] = {
                "serie": detalles["serie"],
                "numero": detalles["numero"]
            }
    elif re.match(r'^[A-Za-z]+-\d+-[A-Za-z]+$', codigo):
        # Forma X-N-Y: 3 bloques letras-digitos-letras. Es la forma de un
        # producto, aunque validar_producto puede rechazarlo si las
        # cantidades exactas no cumplen.
        detalles = validar_producto(codigo)
        resultado["tipo"] = "producto"
        resultado["valido"] = detalles["valido"]
        if detalles["valido"]:
            resultado["detalles"] = {
                "categoria": detalles["categoria"],
                "numero": detalles["numero"],
                "pais": detalles["pais"]
            }

    return resultado
````

````python
# PARTE 3: Procesamiento por lotes

def procesar_lote(codigos: List[str]) -> Dict:
    """
    Procesa múltiples códigos y genera estadísticas agregadas.
    """
    resultado = {
        "total": 0,
        "validos": 0,
        "invalidos": 0,
        "por_tipo": {
            "producto": {"total": 0, "validos": 0},
            "envio": {"total": 0, "validos": 0},
            "empleado": {"total": 0, "validos": 0},
            "factura": {"total": 0, "validos": 0},
            "desconocido": {"total": 0, "validos": 0}
        },
        "detalle": []
    }

    for codigo in codigos:
        validacion = validar_codigo(codigo)
        resultado["detalle"].append(validacion)
        resultado["total"] += 1

        if validacion["valido"]:
            resultado["validos"] += 1
        else:
            resultado["invalidos"] += 1

        tipo = validacion["tipo"]
        resultado["por_tipo"][tipo]["total"] += 1
        if validacion["valido"]:
            resultado["por_tipo"][tipo]["validos"] += 1

    return resultado
````

## Función para Mostrar Resultados

````python
def mostrar_resultado(resultado: Dict) -> None:
    """Muestra el resultado de validación de forma legible."""
    estado = "✓" if resultado["valido"] else "✗"
    print(f"{estado} {resultado['codigo']:<30} | Tipo: {resultado['tipo']:<12}")
    if resultado["valido"] and resultado["detalles"]:
        detalles = ", ".join(f"{k}: {v}" for k, v in resultado["detalles"].items() if v)
        print(f"   └── {detalles}")
````

````python
def mostrar_reporte(reporte: Dict) -> None:
    """Muestra el reporte de procesamiento por lotes."""
    print("=" * 60)
    print("                 REPORTE DE VALIDACIÓN")
    print("=" * 60)
    print(f"\nTotal procesados: {reporte['total']}")
    print(f"Válidos: {reporte['validos']} ({reporte['validos']/reporte['total']*100:.1f}%)")
    print(f"Inválidos: {reporte['invalidos']} ({reporte['invalidos']/reporte['total']*100:.1f}%)")
    
    print("\nDesglose por tipo:")
    print("-" * 40)
    for tipo, stats in reporte["por_tipo"].items():
        if stats["total"] > 0:
            tasa = stats["validos"] / stats["total"] * 100
            print(f"  {tipo.capitalize():<12}: {stats['validos']:>3}/{stats['total']:<3} ({tasa:.0f}% válidos)")
    
    print("\n" + "=" * 60)
````

````python
# BONUS: Funcionalidades extra

import csv
from datetime import date


def sugerir_correccion(codigo: str) -> str:
    """
    Sugiere una correccion para codigos invalidos.

    Reglas aplicadas (en orden):
      1. Quita espacios al inicio/final
      2. Convierte todo a mayusculas
      3. Si la sugerencia ya valida con algun patron, la retorna
      4. Si no, retorna el codigo original sin cambios
    """
    candidato = codigo.strip().upper()

    # Si la version "limpia" ya valida con algun tipo, esa es la sugerencia.
    if validar_codigo(candidato)["valido"]:
        return candidato

    return codigo


def validar_fecha_real(anio: int, mes: int, dia: int) -> bool:
    """Valida si una fecha existe usando datetime (ej: 30 feb -> False)."""
    try:
        date(anio, mes, dia)
        return True
    except ValueError:
        return False


def exportar_resultados(reporte: Dict, archivo: str) -> None:
    """Exporta el detalle de validacion del reporte a un CSV."""
    columnas = ["codigo", "tipo", "valido", "detalles"]

    with open(archivo, "w", encoding="utf-8", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(columnas)

        for item in reporte["detalle"]:
            detalles_str = "; ".join(
                f"{k}={v}" for k, v in item["detalles"].items()
            ) if item["detalles"] else ""
            writer.writerow([
                item["codigo"],
                item["tipo"],
                item["valido"],
                detalles_str
            ])


# Pruebas rapidas del bonus
print("-- sugerir_correccion --")
print(f"  'tec-0001-MX' -> '{sugerir_correccion('tec-0001-MX')}'")
print(f"  '  TEC-0001-MX  ' -> '{sugerir_correccion('  TEC-0001-MX  ')}'")
print(f"  'XXXX' -> '{sugerir_correccion('XXXX')}'")

print("\n-- validar_fecha_real --")
print(f"  2024-02-29 (bisiesto): {validar_fecha_real(2024, 2, 29)}")
print(f"  2025-02-29:            {validar_fecha_real(2025, 2, 29)}")
print(f"  2024-04-31:            {validar_fecha_real(2024, 4, 31)}")

print("\n-- exportar_resultados --")
reporte_demo = procesar_lote(CODIGOS_PRUEBA)
exportar_resultados(reporte_demo, "resultados.csv")
print("  Detalle exportado a 'resultados.csv'")
````
