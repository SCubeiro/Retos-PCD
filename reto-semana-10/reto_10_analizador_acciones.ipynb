## Código Base

````python
import pandas as pd
import numpy as np
from typing import Dict, List, Tuple, Optional
````

````python
# Funciones auxiliares de uso interno para el analizador.

def _fecha_str(fecha) -> str:
    """Formatea una etiqueta del índice (fecha) a 'YYYY-MM-DD' de forma segura."""
    if hasattr(fecha, "strftime"):
        return fecha.strftime("%Y-%m-%d")
    return str(fecha)
````

````python
# PARTE 1: Análisis Estadístico Básico

def estadisticas_basicas(precios: pd.Series) -> Dict:
    """
    Calcula estadísticas descriptivas de los precios usando métodos de Series.

    Los métodos (.min, .max, .mean, ...) ignoran NaN por defecto. Si la serie
    está vacía se devuelven None/0 para no romper el resto del análisis.
    """
    resultado = {
        "precio_actual": None,
        "precio_minimo": None,
        "precio_maximo": None,
        "precio_promedio": None,
        "precio_mediana": None,
        "desviacion_std": None,
        "rango": None,
        "dias_analizados": 0
    }

    # Coacción defensiva: si llegan textos donde se espera un número, se
    # convierten a numérico (lo no convertible queda como NaN y se ignora).
    precios = pd.to_numeric(precios, errors="coerce")
    resultado["dias_analizados"] = int(len(precios))

    if len(precios) == 0:
        return resultado  # serie vacía: no hay nada que calcular

    resultado["precio_actual"] = float(precios.iloc[-1])
    resultado["precio_minimo"] = float(precios.min())
    resultado["precio_maximo"] = float(precios.max())
    resultado["precio_promedio"] = float(precios.mean())
    resultado["precio_mediana"] = float(precios.median())

    # La desviación estándar de un solo dato es NaN; la dejamos en 0.0.
    std = precios.std()
    resultado["desviacion_std"] = float(std) if pd.notna(std) else 0.0
    resultado["rango"] = float(precios.max() - precios.min())

    return resultado
````

````python
def calcular_rendimientos(precios: pd.Series) -> pd.Series:
    """
    Calcula el rendimiento diario porcentual: ((precio_hoy / precio_ayer) - 1) * 100.

    Equivale a .pct_change() * 100. El primer valor siempre es NaN (no hay día
    previo). fill_method=None evita el FutureWarning de pandas 2.x y no rellena
    huecos artificialmente.
    """
    precios = pd.to_numeric(precios, errors="coerce")
    return precios.pct_change(fill_method=None) * 100
````

````python
def analisis_rendimientos(rendimientos: pd.Series) -> Dict:
    """
    Analiza los rendimientos diarios ya calculados.

    'rendimiento_total' se toma como la SUMA de los rendimientos diarios
    (acumulado simple, según la pista .sum() del enunciado). Las fechas de
    mejor/peor día se formatean a 'YYYY-MM-DD' para que el reporte las imprima bien.
    """
    resultado = {
        "rendimiento_total": None,
        "rendimiento_promedio": None,
        "mejor_dia": None,
        "peor_dia": None,
        "dias_positivos": 0,
        "dias_negativos": 0,
        "volatilidad": None
    }

    # El primer rendimiento es NaN; lo descartamos para los cálculos.
    rend = rendimientos.dropna()
    if len(rend) == 0:
        return resultado  # sin datos suficientes

    resultado["rendimiento_total"] = float(rend.sum())
    resultado["rendimiento_promedio"] = float(rend.mean())

    # .idxmax()/.idxmin() devuelven la fecha (índice) del mejor/peor rendimiento.
    resultado["mejor_dia"] = (_fecha_str(rend.idxmax()), float(rend.max()))
    resultado["peor_dia"] = (_fecha_str(rend.idxmin()), float(rend.min()))

    resultado["dias_positivos"] = int((rend > 0).sum())
    resultado["dias_negativos"] = int((rend < 0).sum())

    vol = rend.std()
    resultado["volatilidad"] = float(vol) if pd.notna(vol) else 0.0

    return resultado
````

````python
# PARTE 2: Indicadores Técnicos

def media_movil(precios: pd.Series, ventana: int) -> pd.Series:
    """
    Media móvil simple (SMA) de 'ventana' períodos: .rolling(ventana).mean().

    Los primeros (ventana - 1) valores son NaN porque aún no hay suficiente historia.
    """
    if ventana < 1:
        raise ValueError("La ventana de la media móvil debe ser un entero >= 1.")
    precios = pd.to_numeric(precios, errors="coerce")
    return precios.rolling(window=ventana).mean()
````

````python
def bandas_bollinger(precios: pd.Series, ventana: int = 20, num_std: int = 2) -> Dict:
    """
    Bandas de Bollinger:
    - banda media    = SMA(ventana)
    - banda superior = media + num_std * desviación estándar móvil
    - banda inferior = media - num_std * desviación estándar móvil
    """
    if ventana < 1:
        raise ValueError("La ventana de las bandas de Bollinger debe ser >= 1.")
    precios = pd.to_numeric(precios, errors="coerce")

    media = precios.rolling(window=ventana).mean()
    desv = precios.rolling(window=ventana).std()

    return {
        "banda_superior": media + num_std * desv,
        "banda_media": media,
        "banda_inferior": media - num_std * desv,
    }
````

````python
def detectar_maximos_minimos(precios: pd.Series, ventana: int = 5) -> Dict:
    """
    Detecta máximos y mínimos locales.

    Un punto es máximo local si es el valor más alto de una ventana CENTRADA de
    'ventana' días a cada lado (2*ventana+1 en total); análogo para los mínimos.
    Los extremos de la serie no se evalúan (ventana incompleta -> NaN, no detecta).
    """
    if ventana < 1:
        raise ValueError("La ventana para máximos/mínimos locales debe ser >= 1.")
    precios = pd.to_numeric(precios, errors="coerce")

    tam = 2 * ventana + 1
    max_centrado = precios.rolling(window=tam, center=True).max()
    min_centrado = precios.rolling(window=tam, center=True).min()

    # Un precio es máx/mín local si coincide con el máx/mín de su ventana centrada.
    maximos = precios[precios == max_centrado]
    minimos = precios[precios == min_centrado]

    return {"maximos": maximos, "minimos": minimos}
````

````python
def clasificar_tendencia(precios: pd.Series, ventana: int = 10) -> str:
    """
    Clasifica la tendencia actual en ALCISTA, BAJISTA o LATERAL.

    - ALCISTA: precio actual por encima de la MA y la MA con pendiente positiva.
    - BAJISTA: precio actual por debajo de la MA y la MA con pendiente negativa.
    - LATERAL: cualquier otro caso (incluye datos insuficientes).

    La pendiente de la MA se mide comparando su último valor con el de 'ventana'
    períodos atrás; una tolerancia pequeña evita confundir ruido con tendencia.
    """
    precios = pd.to_numeric(precios, errors="coerce")
    if len(precios) == 0:
        return "LATERAL"

    ma = media_movil(precios, ventana).dropna()
    if len(ma) < 2:
        return "LATERAL"  # no hay MA suficiente para juzgar la pendiente

    precio_actual = precios.iloc[-1]
    ma_actual = ma.iloc[-1]
    # MA de referencia: 'ventana' pasos atrás si existe; si no, la primera disponible.
    ma_referencia = ma.iloc[-ventana] if len(ma) > ventana else ma.iloc[0]
    pendiente = ma_actual - ma_referencia
    tolerancia = abs(ma_actual) * 0.001  # 0.1% del nivel de la MA

    if precio_actual > ma_actual and pendiente > tolerancia:
        return "ALCISTA"
    if precio_actual < ma_actual and pendiente < -tolerancia:
        return "BAJISTA"
    return "LATERAL"
````

````python
# PARTE 3: Sistema de Alertas

def generar_senales_trading(precios: pd.Series, ma_corta: int = 5, ma_larga: int = 20) -> pd.Series:
    """
    Señales por cruce de medias móviles.

    - COMPRA: la MA corta cruza la MA larga de abajo hacia arriba.
    - VENTA:  la MA corta cruza la MA larga de arriba hacia abajo.
    - MANTENER: no hay cruce (o no hay datos suficientes para las MAs).
    """
    precios = pd.to_numeric(precios, errors="coerce")
    ma_c = media_movil(precios, ma_corta)
    ma_l = media_movil(precios, ma_larga)

    # Diferencia entre MAs; su cambio de signo marca el cruce.
    diff = ma_c - ma_l
    diff_prev = diff.shift(1)

    senales = pd.Series("MANTENER", index=precios.index)
    cruce_arriba = (diff_prev <= 0) & (diff > 0)   # COMPRA
    cruce_abajo = (diff_prev >= 0) & (diff < 0)    # VENTA
    senales[cruce_arriba] = "COMPRA"
    senales[cruce_abajo] = "VENTA"
    return senales
````

````python
def alertas_precio(precios: pd.Series, umbral_cambio: float = 5.0) -> List[Dict]:
    """
    Genera alertas cuando el cambio diario supera (en valor absoluto) el umbral.

    Cada alerta: {"fecha": 'YYYY-MM-DD', "tipo": "SUBIDA"/"CAIDA", "cambio": float}.
    Se usa |rendimiento| >= umbral (el umbral es el mínimo para alertar).
    """
    alertas = []
    rendimientos = calcular_rendimientos(precios).dropna()
    significativos = rendimientos[rendimientos.abs() >= umbral_cambio]

    for fecha, cambio in significativos.items():
        alertas.append({
            "fecha": _fecha_str(fecha),
            "tipo": "SUBIDA" if cambio > 0 else "CAIDA",
            "cambio": float(cambio),
        })
    return alertas
````

````python
def clasificar_volatilidad(rendimientos: pd.Series) -> str:
    """
    Clasifica la volatilidad según la desviación estándar de los rendimientos:
    BAJA (< 1%), MEDIA ([1, 3)%), ALTA ([3, 5)%) y MUY ALTA (>= 5%).
    """
    vol = rendimientos.std()
    if pd.isna(vol):
        return "N/A"  # sin datos suficientes para estimar volatilidad
    if vol < 1:
        return "BAJA"
    if vol < 3:
        return "MEDIA"
    if vol < 5:
        return "ALTA"
    return "MUY ALTA"
````

````python
def generar_reporte_completo(precios: pd.Series, nombre_accion: str) -> Dict:
    """
    Reporte completo que integra todas las funciones anteriores.
    """
    reporte = {
        "nombre": nombre_accion,
        "periodo": {},
        "estadisticas": {},
        "rendimientos": {},
        "tendencia": None,
        "volatilidad": None,
        "senal_actual": None,
        "alertas_recientes": []
    }

    # Supuesto: precios en orden cronológico. Ordenamos por índice por robustez.
    precios = precios.sort_index()
    if len(precios) == 0:
        return reporte  # sin datos no hay reporte que generar

    rendimientos = calcular_rendimientos(precios)

    reporte["periodo"] = {
        "inicio": _fecha_str(precios.index[0]),
        "fin": _fecha_str(precios.index[-1]),
        "dias": int(len(precios)),
    }
    reporte["estadisticas"] = estadisticas_basicas(precios)
    reporte["rendimientos"] = analisis_rendimientos(rendimientos)
    reporte["tendencia"] = clasificar_tendencia(precios)
    reporte["volatilidad"] = clasificar_volatilidad(rendimientos)

    senales = generar_senales_trading(precios)
    reporte["senal_actual"] = senales.iloc[-1] if len(senales) else "MANTENER"
    reporte["alertas_recientes"] = alertas_precio(precios)

    return reporte
````

````python
def mostrar_reporte(reporte: Dict) -> None:
    """Muestra el reporte de forma legible."""
    print("=" * 70)
    print(f"           REPORTE DE ANÁLISIS: {reporte['nombre']}")
    print("=" * 70)
    
    # Período
    periodo = reporte.get('periodo', {})
    print(f"\n📅 PERÍODO DE ANÁLISIS")
    print("-" * 40)
    print(f"Inicio: {periodo.get('inicio', 'N/A')}")
    print(f"Fin: {periodo.get('fin', 'N/A')}")
    print(f"Días analizados: {periodo.get('dias', 'N/A')}")
    
    # Estadísticas
    stats = reporte.get('estadisticas', {})
    print(f"\n📊 ESTADÍSTICAS DE PRECIO")
    print("-" * 40)
    print(f"Precio actual:  ${stats.get('precio_actual', 0):,.2f}")
    print(f"Precio mínimo:  ${stats.get('precio_minimo', 0):,.2f}")
    print(f"Precio máximo:  ${stats.get('precio_maximo', 0):,.2f}")
    print(f"Precio promedio: ${stats.get('precio_promedio', 0):,.2f}")
    
    # Rendimientos
    rend = reporte.get('rendimientos', {})
    print(f"\n📈 RENDIMIENTO")
    print("-" * 40)
    print(f"Rendimiento total: {rend.get('rendimiento_total', 0):+.2f}%")
    print(f"Rendimiento promedio diario: {rend.get('rendimiento_promedio', 0):+.3f}%")
    if rend.get('mejor_dia'):
        print(f"Mejor día: {rend['mejor_dia'][0]} ({rend['mejor_dia'][1]:+.2f}%)")
    if rend.get('peor_dia'):
        print(f"Peor día: {rend['peor_dia'][0]} ({rend['peor_dia'][1]:+.2f}%)")
    print(f"Días positivos: {rend.get('dias_positivos', 0)}")
    print(f"Días negativos: {rend.get('dias_negativos', 0)}")
    
    # Indicadores
    print(f"\n🎯 INDICADORES")
    print("-" * 40)
    print(f"Tendencia: {reporte.get('tendencia', 'N/A')}")
    print(f"Volatilidad: {reporte.get('volatilidad', 'N/A')}")
    print(f"Señal actual: {reporte.get('senal_actual', 'N/A')}")
    
    # Alertas
    alertas = reporte.get('alertas_recientes', [])
    if alertas:
        print(f"\n⚠️ ALERTAS RECIENTES")
        print("-" * 40)
        for alerta in alertas[-5:]:  # Últimas 5
            emoji = "🔺" if alerta['tipo'] == 'SUBIDA' else "🔻"
            print(f"{emoji} {alerta['fecha']}: {alerta['tipo']} de {alerta['cambio']:+.2f}%")
    
    print("\n" + "=" * 70)
````

````python
def visualizar_precios_texto(precios: pd.Series, ancho: int = 50) -> None:
    """Visualización simple de precios en texto (ASCII chart)."""
    min_precio = precios.min()
    max_precio = precios.max()
    rango = max_precio - min_precio
    
    print(f"\nGráfico de precios: {precios.name}")
    print(f"Max: ${max_precio:.2f}")
    print("-" * (ancho + 10))
    
    # Mostrar cada 3 días para no saturar
    for fecha, precio in precios.iloc[::3].items():
        posicion = int((precio - min_precio) / rango * ancho) if rango > 0 else ancho // 2
        barra = " " * posicion + "█"
        fecha_str = fecha.strftime('%m/%d') if hasattr(fecha, 'strftime') else str(fecha)[:5]
        print(f"{fecha_str} |{barra}")
    
    print("-" * (ancho + 10))
    print(f"Min: ${min_precio:.2f}")
````
